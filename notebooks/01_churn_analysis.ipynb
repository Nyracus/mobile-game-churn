## 2. Model selection protocol

All model-family, hyperparameter, and threshold decisions use the training split only. Five-fold stratified randomized search selects the model by ROC-AUC, and out-of-fold probabilities select the churn-F1 threshold. The development set remains an untouched final holdout.

```python
selection = pd.read_csv(REPORTS / 'model_selection_summary.csv')
display(selection[['model', 'cv_roc_auc', 'cv_roc_auc_std', 'dev_roc_auc', 'dev_pr_auc', 'dev_f1']])
display(Image(filename=str(FIGURES / 'tuned_model_comparison.png'), width=900))
display(Image(filename=str(FIGURES / 'threshold_selection.png'), width=750))
```
